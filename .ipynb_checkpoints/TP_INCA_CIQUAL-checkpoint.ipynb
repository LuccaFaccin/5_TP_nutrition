
# Partie B — Recherche

## B1. Rechercher un aliment

Compléter la fonction `rechercher_aliment(texte)`.

La fonction doit retourner les aliments dont le libellé contient le texte recherché, **sans tenir compte des majuscules/minuscules**.

````python
def rechercher_aliment(texte):
    """Retourne les aliments dont le libellé contient texte (insensible à la casse)."""

    texte = texte.lower()

    resultats = aliments[
        aliments["libelle"].str.lower().str.contains(texte, na=False, regex=False)
    ]

    return resultats
````

````python
# Exemples
display(rechercher_aliment("pain"))
display(rechercher_aliment("POMME"))
````

````python
# Vérification
r = rechercher_aliment("pain")
assert len(r) == 1
assert r.iloc[0]["code_aliment"] == 1001
assert len(rechercher_aliment("POMME")) == 1
print("B1 : OK")
````

## B2. Composition d'un aliment

Compléter la fonction `composition_aliment(code_aliment)`.

Le résultat doit contenir les colonnes :

- `code_const`
- `nom`
- `teneur`
- `unite`

Tester avec le pain (`1001`).

````python
def composition_aliment(code_aliment):
    """Retourne la composition d'un aliment, avec nom et unité du constituant."""

    # Sélectionner les données de l'aliment
    resultat = composition[composition["code_aliment"] == code_aliment]

    # Ajouter les informations des constituants
    resultat = resultat.merge(
        constituants,
        on="code_const",
        how="left"
    )

    # Sélectionner les colonnes demandées
    resultat = resultat[["code_const", "nom", "teneur", "unite"]]

    return resultat
````

## C1. Apport d'un aliment

Compléter la fonction `apport_aliment(code_aliment, quantite_g)`.

Le résultat doit contenir une colonne `apport`.

Pour 60 g de pain, vérifier notamment que l'apport en protéines est :

```text
8,5 × 60 / 100 = 5,1 g
```

````python
def apport_aliment(code_aliment, quantite_g):
    """Calcule les apports pour une quantité donnée en grammes."""

    # Récupérer la composition nutritionnelle de l'aliment
    resultat = composition_aliment(code_aliment).copy()

    # Calculer les apports pour la quantité consommée
    resultat["apport"] = resultat["teneur"] * quantite_g / 100

    return resultat
````

````python
apport_aliment(1001, 60)
````

````python
# Vérification
r = apport_aliment(1001, 60)
valeur = float(r.loc[r["code_const"] == 2, "apport"].iloc[0])
assert abs(valeur - 5.1) < 1e-9
print("C1 : OK")
````

## C2. Apports d'un individu

Compléter la fonction `apports_individu(id_individu)`.

La fonction doit :

1. sélectionner les consommations de l'individu ;
2. retrouver les aliments ;
3. retrouver leur composition ;
4. calculer l'apport correspondant à chaque consommation ;
5. additionner les apports pour chaque constituant.

*Dans cette partie, vous pouvez par exemple réutiliser `apport_aliment()` dans une boucle sur les consommations.*

````python
def apports_individu(id_individu):
    """Calcule les apports nutritionnels totaux d'un individu."""

    # 1. Sélectionner les consommations de l'individu
    conso_individu = consommations[
        consommations["id_individu"] == id_individu
    ]

    # 2. Créer une liste pour stocker les apports
    liste_apports = []

    # 3. Parcourir les consommations
    for i, ligne in conso_individu.iterrows():

        code_aliment = ligne["code_aliment"]
        quantite_g = ligne["quantite_g"]

        # 4. Calculer les apports de chaque aliment
        apport = apport_aliment(code_aliment, quantite_g)

        liste_apports.append(apport)

    # 5. Regrouper tous les résultats
    if len(liste_apports) == 0:
        return pd.DataFrame(columns=["code_const", "nom", "unite", "apport"])

    tous_apports = pd.concat(liste_apports, ignore_index=True)

    # 6. Additionner les apports par constituant
    resultat = tous_apports.groupby(
        ["code_const", "nom", "unite"],
        as_index=False
    )["apport"].sum()

    return resultat
````

# Partie D — Les jointures

Réécrire `apports_individu()` en utilisant explicitement `merge()`.

On cherche à obtenir conceptuellement :

```text
consommations
      │
      ├── merge aliments
      │
      └── merge composition
                │
                └── merge constituants
```

````python
def apports_individu_merge(id_individu):
    """Calcule les apports nutritionnels totaux d'un individu à l'aide de merge()."""

    # 1. Sélectionner les consommations de l'individu
    resultat = consommations[
        consommations["id_individu"] == id_individu
    ].copy()

    # 2. Jointure avec les aliments
    resultat = resultat.merge(
        aliments,
        on="code_aliment",
        how="inner"
    )

    # 3. Jointure avec la composition nutritionnelle
    resultat = resultat.merge(
        composition,
        on="code_aliment",
        how="inner"
    )

    # 4. Jointure avec les constituants
    resultat = resultat.merge(
        constituants,
        on="code_const",
        how="inner"
    )

    # 5. Calculer les apports nutritionnels
    resultat["apport"] = (
        resultat["teneur"] * resultat["quantite_g"] / 100
    )

    # 6. Additionner les apports par constituant
    resultat = resultat.groupby(
        ["code_const", "nom", "unite"],
        as_index=False
    )["apport"].sum()

    return resultat
````

````python
apports_individu_merge(1)
````

````python
# Vérification : les deux versions doivent donner les mêmes totaux.
# Adapter si nécessaire le nom des colonnes ("nom", "apport") à votre implémentation.
v1 = apports_individu(1).set_index("nom")["apport"].sort_index()
v2 = apports_individu_merge(1).set_index("nom")["apport"].sort_index()
assert (v1 - v2).abs().max() < 1e-9
print("D : OK")
````

**Questions**

1. Quelle colonne permet la jointure avec `aliments` ?
2. Quelle colonne permet la jointure avec `composition` ?
3. Pourquoi `composition` augmente-t-elle fortement le nombre de lignes ?
4. Que représente une ligne du DataFrame final ?

> **Réponses :**
>
> 1. La jointure avec la table aliments se fait à l'aide de la colonne code_aliment, qui permet d'identifier chaque aliment.
> 2. La jointure avec la table composition utilise également la colonne code_aliment, afin d'associer chaque aliment à ses constituants nutritionnels.
> 3. La table composition contient plusieurs lignes pour chaque aliment, car un aliment possède plusieurs constituants nutritionnels. Lors de la jointure, chaque consommation est donc associée à plusieurs lignes, ce qui augmente le nombre total de lignes.
> Avant l'agrégation, chaque ligne représente un constituant nutritionnel associé à un aliment consommé par un individu, avec sa teneur et la quantité consommée.
Après l'agrégation avec groupby(), chaque ligne représente l'apport nutritionnel total de l'individu pour un constituant donné.

# Partie E — Analyse par repas

Compléter la fonction `apports_par_repas(id_individu)`.

La fonction doit produire les apports par repas.

Exemple de structure attendue :

```text
repas              Energie    Proteines    Lipides
petit_dejeuner     ...
collation          ...
dejeuner           ...
diner              ...
```

*Indice : penser à `groupby` puis à `pivot_table` (ou `unstack`).*

````python
def apports_par_repas(id_individu):
    """Calcule les apports nutritionnels par repas pour un individu."""

    # 1. Sélectionner les consommations de l'individu
    resultat = consommations[
        consommations["id_individu"] == id_individu
    ].copy()

    # 2. Jointure avec la composition
    resultat = resultat.merge(
        composition,
        on="code_aliment",
        how="inner"
    )

    # 3. Jointure avec les constituants
    resultat = resultat.merge(
        constituants,
        on="code_const",
        how="inner"
    )

    # 4. Calculer les apports
    resultat["apport"] = (
        resultat["teneur"] * resultat["quantite_g"] / 100
    )

    # 5. Regrouper les apports par repas et par nutriment
    resultat = resultat.groupby(
        ["repas", "nom"],
        as_index=False
    )["apport"].sum()

    # 6. Transformer les nutriments en colonnes
    resultat = resultat.pivot_table(
        index="repas",
        columns="nom",
        values="apport",
        aggfunc="sum",
        fill_value=0
    )

    return resultat
````

# Partie G — Qualité des données

Le fichier `consommations.csv` constitue un jeu simplifié.

Imaginez maintenant que certaines données réelles puissent contenir :

- un aliment sans code correspondant dans la table des aliments ;
- une quantité nulle ;
- une quantité négative ;
- un code constituant absent ;
- une teneur manquante ;
- plusieurs lignes identiques.

Écrire une fonction `controler_donnees()` qui détecte **au moins trois catégories** d'anomalies.

*Conseil : retourner par exemple un dictionnaire `{catégorie d'anomalie: DataFrame des lignes concernées}`.*

````python
def controler_donnees(consommations_test=None, aliments_test=None, composition_test=None):
    """Détecte des anomalies dans les données."""

    if consommations_test is None:
        consommations_test = consommations

    if aliments_test is None:
        aliments_test = aliments

    if composition_test is None:
        composition_test = composition

    anomalies = {}

    # 1. Codes alimentaires inexistants
    anomalies["aliments_inexistants"] = consommations_test[
        ~consommations_test["code_aliment"].isin(aliments_test["code_aliment"])
    ]

    # 2. Quantités manquantes
    anomalies["quantites_manquantes"] = consommations_test[
        consommations_test["quantite_g"].isna()
    ]

    # 3. Quantités négatives
    anomalies["quantites_negatives"] = consommations_test[
        consommations_test["quantite_g"] < 0
    ]

    # 4. Teneurs manquantes
    anomalies["teneurs_manquantes"] = composition_test[
        composition_test["teneur"].isna()
    ]

    # 5. Doublons
    anomalies["doublons"] = consommations_test[
        consommations_test.duplicated(keep=False)
    ]

    return anomalies
````

# Partie H — Défi algorithmique

Écrire une fonction `aliment_plus_proteine(id_individu)`.

Elle doit retourner l'aliment ayant fourni la plus grande quantité de protéines à l'individu.

**Contrainte :** utiliser au moins une jointure et une agrégation.

````python
def aliment_plus_proteine(id_individu):
    """Retourne l'aliment ayant fourni le plus de protéines à l'individu."""

    # 1. Sélectionner les consommations de l'individu
    resultat = consommations[
        consommations["id_individu"] == id_individu
    ].copy()

    # 2. Jointure avec la composition nutritionnelle
    resultat = resultat.merge(
        composition,
        on="code_aliment",
        how="inner"
    )

    # 3. Jointure avec les constituants
    resultat = resultat.merge(
        constituants,
        on="code_const",
        how="inner"
    )

    # 4. Sélectionner uniquement les protéines
    resultat = resultat[
        resultat["nom"].str.lower().isin(["protéines", "proteines"])
    ].copy()

    # 5. Calculer les apports en protéines
    resultat["apport"] = (
        resultat["teneur"] * resultat["quantite_g"] / 100
    )

    # 6. Additionner les apports par aliment
    resultat = resultat.groupby(
        "code_aliment",
        as_index=False
    )["apport"].sum()

    # 7. Jointure avec les noms des aliments
    resultat = resultat.merge(
        aliments,
        on="code_aliment",
        how="left"
    )

    # 8. Identifier l'aliment avec le plus de protéines
    if resultat.empty:
        return None

    aliment_max = resultat.loc[
        resultat["apport"].idxmax()
    ]

    return aliment_max
````
